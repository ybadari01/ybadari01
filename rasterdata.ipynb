Introduction to GDAL
The Geospatial Data Abstraction Library is a translator library for raster and vector geospatial data formats  
GDAL is a computer software libraray fo reading amd writing raster and vector geospatial data format
Geospatial data abstarctiobn libraray presents a single raster abstract data model and single vectopr abstract data model to the calling application for all supported formats
GDAL also comes with a variety of useful command line utilites for data translation and processing

Advatnages of using GDAL
Free and open source (https://github.com/OSGeo/gdal)
support for over 80+ Image formats and map projections
Command line as well as c/c++/python/R/Java API
Used extensively by worlds large geospatia data services.
Extensive test suite and active developer community
GDAL also includes extensive support for vector datasets
